```python
# ==========================================
# Day 8 - Pandas Indexing, Selection & Filtering
# ==========================================

import pandas as pd

# -----------------------------
# Create a DataFrame
# -----------------------------

data = {
    "Name": ["Rahman", "Ali", "Sara", "Ahmed", "Fatima"],
    "Age": [20, 22, 19, 23, 21],
    "Marks": [85, 78, 92, 88, 95],
    "City": ["Raichur", "Bangalore", "Hyderabad", "Delhi", "Mumbai"]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n")
print(df)

# -----------------------------
# Selecting Columns
# -----------------------------

print("\nSingle Column:")
print(df["Name"])

print("\nMultiple Columns:")
print(df[["Name", "Marks"]])

# -----------------------------
# Selecting Rows
# -----------------------------

print("\nFirst Three Rows:")
print(df[0:3])

# -----------------------------
# Using loc[]
# -----------------------------

print("\nUsing loc[]:")
print(df.loc[1])

print("\nRows 1 to 3:")
print(df.loc[1:3])

# -----------------------------
# Using iloc[]
# -----------------------------

print("\nUsing iloc[]:")
print(df.iloc[2])

print("\nRows 0 to 2:")
print(df.iloc[0:3])

# -----------------------------
# Boolean Filtering
# -----------------------------

print("\nStudents with Marks > 85")
print(df[df["Marks"] > 85])

print("\nStudents Age > 20")
print(df[df["Age"] > 20])

# -----------------------------
# Multiple Conditions
# -----------------------------

print("\nMarks > 80 and Age > 20")
print(df[(df["Marks"] > 80) & (df["Age"] > 20)])

# -----------------------------
# Sorting
# -----------------------------

print("\nSort by Marks (Ascending)")
print(df.sort_values("Marks"))

print("\nSort by Marks (Descending)")
print(df.sort_values("Marks", ascending=False))

# -----------------------------
# Reset Index
# -----------------------------

print("\nReset Index")
print(df.reset_index(drop=True))
```

```python
import pandas as pd
import numpy as np

# -------------------------------------
# Day 9 - Missing Values & Data Cleaning
# -------------------------------------

data = {
    "Name": ["Rahman", "Ali", "Sara", "Ahmed", "Fatima", "Ali"],
    "Age": [20, np.nan, 19, 23, np.nan, np.nan],
    "Marks": [85, 78, np.nan, 88, 95, 78],
    "City": ["Raichur", "Bangalore", "Hyderabad", None, "Mumbai", "Bangalore"]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)

# Missing Values
print("\nMissing Values")
print(df.isnull())

print("\nMissing Values Count")
print(df.isnull().sum())

# Fill Missing Values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna("Unknown")

print("\nAfter Filling Missing Values")
print(df)

# Duplicate Rows
print("\nDuplicate Rows")
print(df.duplicated())

df = df.drop_duplicates()

print("\nAfter Removing Duplicates")
print(df)

# Rename Column
df = df.rename(columns={"Marks": "Score"})

print("\nRenamed Columns")
print(df)

# Data Summary
print("\nData Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())
```

```python
import pandas as pd

# Day 10 - Data Transformation

data = {
    "Name": ["Rahman", "Ali", "Sara", "Ahmed"],
    "Gender": ["M", "M", "F", "M"],
    "Salary": [50000, 60000, 55000, 70000],
    "City": ["BLR", "HYD", "DEL", "BLR"]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)

# apply()
df["Salary"] = df["Salary"].apply(lambda x: x + 5000)

# map()
df["Gender"] = df["Gender"].map({"M":"Male","F":"Female"})

# replace()
df["City"] = df["City"].replace({
    "BLR":"Bangalore",
    "HYD":"Hyderabad",
    "DEL":"Delhi"
})

# String Operations
df["Upper_Name"] = df["Name"].str.upper()
df["Name_Length"] = df["Name"].str.len()

# New Column
df["Bonus"] = df["Salary"] * 0.10

print("\nUpdated DataFrame")
print(df)
```

```python
import pandas as pd

# Day 11 - GroupBy Basics

data = {
    "Department":["IT","HR","IT","Finance","HR","Finance","IT"],
    "Employee":["Rahman","Ali","Sara","Ahmed","Fatima","John","David"],
    "Salary":[50000,45000,65000,70000,48000,72000,60000]
}

df = pd.DataFrame(data)

print(df)

print("\nGroup Size")
print(df.groupby("Department").size())

print("\nAverage Salary")
print(df.groupby("Department")["Salary"].mean())

print("\nMaximum Salary")
print(df.groupby("Department")["Salary"].max())

print("\nMinimum Salary")
print(df.groupby("Department")["Salary"].min())

print("\nTotal Salary")
print(df.groupby("Department")["Salary"].sum())

print("\nCount")
print(df.groupby("Department")["Salary"].count())
```

```python
import pandas as pd

# Day 12 - Advanced GroupBy

data = {
    "Department":["IT","HR","IT","Finance","HR","Finance","IT"],
    "Employee":["Rahman","Ali","Sara","Ahmed","Fatima","John","David"],
    "Salary":[50000,45000,65000,70000,48000,72000,60000],
    "Experience":[2,4,3,6,5,7,2]
}

df = pd.DataFrame(data)

print(df)

print("\nMultiple Aggregations")
print(
    df.groupby("Department").agg({
        "Salary":["mean","max","min","sum"],
        "Experience":["mean","max"]
    })
)

print("\nDescribe")
print(df.groupby("Department").describe())

print("\nGet Group - IT")
print(df.groupby("Department").get_group("IT"))

print("\nIterating Groups")

for dept, group in df.groupby("Department"):
    print("\nDepartment:", dept)
    print(group)
```

```python
import pandas as pd

students = pd.DataFrame({
    "ID":[1,2,3,4],
    "Name":["Rahman","Ali","Sara","Ahmed"]
})

marks = pd.DataFrame({
    "ID":[1,2,3,5],
    "Marks":[91,85,78,88]
})

print("Students")
print(students)

print("\nMarks")
print(marks)

# Inner Merge
print("\nInner Merge")
print(pd.merge(students, marks, on="ID"))

# Left Join
print("\nLeft Join")
print(pd.merge(students, marks, on="ID", how="left"))

# Right Join
print("\nRight Join")
print(pd.merge(students, marks, on="ID", how="right"))

# Outer Join
print("\nOuter Join")
print(pd.merge(students, marks, on="ID", how="outer"))

# Concatenation

df1 = pd.DataFrame({"A":[1,2],"B":[3,4]})
df2 = pd.DataFrame({"A":[5,6],"B":[7,8]})

print("\nVertical Concat")
print(pd.concat([df1,df2]))

print("\nHorizontal Concat")
print(pd.concat([df1,df2],axis=1))
```

```python
import pandas as pd

data = {
    "Department":["IT","IT","HR","HR","Finance","Finance"],
    "City":["Bangalore","Delhi","Bangalore","Delhi","Delhi","Bangalore"],
    "Gender":["M","F","F","M","M","F"],
    "Salary":[50000,60000,45000,47000,65000,70000]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)

print("\nMean Salary by Department and City")
pivot = df.pivot_table(
    values="Salary",
    index="Department",
    columns="City",
    aggfunc="mean"
)

print(pivot)

print("\nSalary Sum")
print(df.pivot_table(values="Salary",
                     index="Department",
                     aggfunc="sum"))

print("\nCrosstab")
print(pd.crosstab(df["Department"],df["Gender"]))

print("\nMultiIndex")
multi_df = df.set_index(["Department","City"])
print(multi_df)

print("\nAccess Single Row")
print(multi_df.loc[("IT","Delhi")])

print("\nReset Index")
print(multi_df.reset_index())
```

```python
import pandas as pd

data = {
    "Date": [
        "2026-07-20",
        "2026-07-21",
        "2026-07-22",
        "2026-07-23",
        "2026-07-24"
    ],
    "Sales": [500,650,420,780,620]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()

print(df)

print("\nSorted Data")
print(df.sort_values("Date"))
```

```python
import pandas as pd

# ==========================
# Load Dataset
# ==========================
df = pd.read_csv("batsman_runs_ipl.csv")

# ==========================
# Basic Information
# ==========================
print("\n===== HEAD =====")
print(df.head())

print("\n===== TAIL =====")
print(df.tail())

print("\n===== SHAPE =====")
print(df.shape)

print("\n===== COLUMNS =====")
print(df.columns)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== INFO =====")
print(df.info())

# ==========================
# Column Selection
# ==========================
print("\n===== SINGLE COLUMN =====")
print(df["batter"])

print("\n===== MULTIPLE COLUMNS =====")
print(df[["batter", "batsman_run"]])

# ==========================
# Row Selection
# ==========================
print("\n===== FIRST 5 ROWS =====")
print(df[0:5])

print("\n===== LOC =====")
print(df.loc[0:5])

print("\n===== ILOC =====")
print(df.iloc[0:5])

# ==========================
# Filtering
# ==========================
print("\n===== RUNS > 20 =====")
print(df[df["batsman_run"] > 20])

print("\n===== RUNS BETWEEN 10 & 50 =====")
print(df[(df["batsman_run"] >= 10) & (df["batsman_run"] <= 50)])

# ==========================
# Sorting
# ==========================
print("\n===== SORT BY RUNS =====")
print(df.sort_values("batsman_run", ascending=False))

print("\n===== SORT BY BATTER =====")
print(df.sort_values("batter"))

# ==========================
# Unique Values
# ==========================
print("\n===== UNIQUE BATTERS =====")
print(df["batter"].unique())

print("\n===== NUMBER OF UNIQUE BATTERS =====")
print(df["batter"].nunique())

print("\n===== VALUE COUNTS =====")
print(df["batter"].value_counts())

# ==========================
# Statistics
# ==========================
print("\n===== DESCRIBE =====")
print(df.describe())

print("\n===== MEAN =====")
print(df["batsman_run"].mean())

print("\n===== MEDIAN =====")
print(df["batsman_run"].median())

print("\n===== MAX =====")
print(df["batsman_run"].max())

print("\n===== MIN =====")
print(df["batsman_run"].min())

# ==========================
# Missing Values
# ==========================
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Uncomment if required
# print(df.dropna())
# print(df.fillna(0))

# ==========================
# Duplicate Rows
# ==========================
print("\n===== DUPLICATES =====")
print(df.duplicated().sum())

# Uncomment if required
# print(df.drop_duplicates())

# ==========================
# New Column
# ==========================
df["Double_Runs"] = df["batsman_run"] * 2

print("\n===== NEW COLUMN =====")
print(df.head())

# ==========================
# Apply Function
# ==========================
df["Half_Runs"] = df["batsman_run"].apply(lambda x: x // 2)

print("\n===== APPLY FUNCTION =====")
print(df.head())

# ==========================
# Rename Column
# ==========================
renamed_df = df.rename(columns={"batsman_run": "Runs"})

print("\n===== RENAME COLUMN =====")
print(renamed_df.head())

# ==========================
# GroupBy
# ==========================
print("\n===== TOTAL RUNS BY BATTER =====")
print(df.groupby("batter")["batsman_run"].sum())

print("\n===== AVERAGE RUNS =====")
print(df.groupby("batter")["batsman_run"].mean())

print("\n===== MAX RUNS =====")
print(df.groupby("batter")["batsman_run"].max())

print("\n===== COUNT =====")
print(df.groupby("batter")["batsman_run"].count())

# ==========================
# Export CSV
# ==========================
df.to_csv("output.csv", index=False)

print("\n✅ Pandas Fundamentals Completed Successfully!")
```
