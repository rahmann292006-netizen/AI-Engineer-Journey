```python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.figure(figsize=(6,4))
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1,11)
y = x**2

# Figure Size
plt.figure(figsize=(8,5))

# Customized Line Plot
plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8,
    label="y = x²"
)

# Title & Labels
plt.title("Customized Line Plot", fontsize=16)
plt.xlabel("X Values", fontsize=12)
plt.ylabel("Y Values", fontsize=12)

# Grid
plt.grid(True)

# Legend
plt.legend()

# Axis Limits
plt.xlim(1,10)
plt.ylim(0,110)

# Annotation
plt.annotate(
    "Highest Point",
    xy=(10,100),
    xytext=(7,80),
    arrowprops=dict(arrowstyle="->")
)

# Save Figure
plt.savefig("custom_plot.png")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# Load Dataset
# ==========================================

df = pd.read_csv(r"C:\Users\Abdul Rahman\Downloads\deliveries (1).csv")

print("="*60)
print("First 5 Rows")
print(df.head())

print("="*60)
print("Columns")
print(df.columns.tolist())

print("="*60)
print(df.info())

print("="*60)
print(df.describe())

# ==========================================
# Detect Column Names Automatically
# ==========================================

# Batter Column
if "batter" in df.columns:
    batter_col = "batter"
elif "batsman" in df.columns:
    batter_col = "batsman"
else:
    raise Exception("Batter column not found!")

# Runs Column
if "batsman_run" in df.columns:
    runs_col = "batsman_run"
elif "batsman_runs" in df.columns:
    runs_col = "batsman_runs"
elif "total_runs" in df.columns:
    runs_col = "total_runs"
else:
    raise Exception("Runs column not found!")

print(f"\nUsing Batter Column : {batter_col}")
print(f"Using Runs Column   : {runs_col}")

# ==========================================
# Top 10 Batters
# ==========================================

top_batters = (
    df.groupby(batter_col)[runs_col]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("\nTop 10 Batters")
print(top_batters)

plt.figure(figsize=(10,5))

plt.bar(top_batters.index, top_batters.values)

plt.title("Top 10 Batters by Runs")
plt.xlabel("Batter")
plt.ylabel("Runs")

plt.xticks(rotation=45)

plt.grid(axis="y")

plt.tight_layout()

plt.show()

# ==========================================
# Histogram
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df[runs_col], bins=10)

plt.title("Distribution of Runs")

plt.xlabel("Runs")

plt.ylabel("Frequency")

plt.grid(True)

plt.show()

# ==========================================
# Scatter Plot
# ==========================================

plt.figure(figsize=(10,5))

plt.scatter(df.index, df[runs_col], s=5)

plt.title("Runs vs Ball Index")

plt.xlabel("Ball Number")

plt.ylabel("Runs")

plt.grid(True)

plt.show()

# ==========================================
# Pie Chart
# ==========================================

top5 = top_batters.head(5)

plt.figure(figsize=(7,7))

plt.pie(
    top5.values,
    labels=top5.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Batters")

plt.show()

# ==========================================
# Subplots
# ==========================================

fig, ax = plt.subplots(1,2, figsize=(14,5))

ax[0].bar(top_batters.index, top_batters.values)
ax[0].set_title("Top 10 Batters")
ax[0].tick_params(axis="x", rotation=45)

ax[1].hist(df[runs_col], bins=10)
ax[1].set_title("Runs Distribution")

plt.tight_layout()

plt.show()

# ==========================================
# Save Figure
# ==========================================

plt.figure(figsize=(10,5))

plt.bar(top_batters.index, top_batters.values)

plt.title("Top 10 Batters")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("top10_batters.png")

print("\n✅ Figure saved as top10_batters.png")

print("\n🎉 Day 23 Completed Successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("car_price.csv")

# ==========================
# Dataset Overview
# ==========================

print("First 5 Rows")
print(df.head())

print("\nShape")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nMissing Values")
print(df.isnull().sum())

# ==========================
# Numerical Summary
# ==========================

print("\nSummary Statistics")
print(df.describe())

# ==========================
# Car Price Distribution
# ==========================

plt.figure(figsize=(8,5))
plt.hist(df["price"], bins=20, edgecolor="black")
plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

# ==========================
# Fuel Type Count
# ==========================

fuel_counts = df["fueltype"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(fuel_counts.index, fuel_counts.values)
plt.title("Fuel Type Distribution")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.show()

# ==========================
# Transmission Count
# ==========================

transmission_counts = df["drivewheel"].value_counts()

plt.figure(figsize=(6,4))
plt.pie(
    transmission_counts.values,
    labels=transmission_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Transmission Distribution")
plt.show()

# ==========================
# Year vs Price
# ==========================

plt.figure(figsize=(8,5))
# choose a valid x (Year if present, else try to extract from CarName, else use index)
if "Year" in df.columns:
    x = df["Year"]
elif "year" in df.columns:
    x = df["year"]
else:
    if "CarName" in df.columns:
        yrs = df["CarName"].str.extract(r"(\d{4})")[0]
        x = pd.to_numeric(yrs, errors='coerce').fillna(df.index.to_series()).astype(float)
    else:
        x = df.index

# choose price column (lowercase 'price' used in this dataset)
y = df["price"] if "price" in df.columns else df.get("Price", df.iloc[:, -1])

plt.scatter(x, y)
plt.title("Year vs Price")
plt.xlabel("Year")
plt.ylabel("Price")
plt.grid(True)
plt.show()

# ==========================
# Top 10 Expensive Cars
# ==========================

top10 = df.sort_values(by="price", ascending=False).head(10).copy()
top10 = top10.rename(columns={"CarName": "Car_Name", "price": "Price"})

plt.figure(figsize=(10,5))
plt.bar(top10["Car_Name"], top10["Price"])
plt.title("Top 10 Most Expensive Cars")
plt.xlabel("Car")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
