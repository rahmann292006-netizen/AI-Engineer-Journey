# ===============================
# 1. Line Plot
# ===============================

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.figure(figsize=(6,4))
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

 2. Line Plot with Marker
 

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.figure(figsize=(6,4))
plt.plot(x, y, marker="o")
plt.title("Line Plot with Markers")
plt.grid(True)
plt.show()
```

# ===============================
# 3. Multiple Lines
# ===============================

```python
y2 = [1,4,9,16,25]

plt.figure(figsize=(6,4))
plt.plot(x, y, label="2x")
plt.plot(x, y2, label="x²")
plt.legend()
plt.title("Multiple Lines")
plt.grid(True)
plt.show()
```

# ===============================
# 4. Bar Chart
# ===============================

```python
students = ["A","B","C","D"]
marks = [75,82,65,90]

plt.figure(figsize=(6,4))
plt.bar(students, marks)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

# ===============================
# 5. Horizontal Bar Chart
# ===============================

```python
plt.figure(figsize=(6,4))
plt.barh(students, marks)
plt.title("Horizontal Bar")
plt.show()
```

# ===============================
# 6. Histogram
# ===============================

```python
data = np.random.randint(1,100,100)

plt.figure(figsize=(6,4))
plt.hist(data, bins=10)
plt.title("Histogram")
plt.show()
```

# ===============================
# 7. Scatter Plot
# ===============================

```python
x = np.random.randint(1,100,50)
y = np.random.randint(1,100,50)

plt.figure(figsize=(6,4))
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

# ===============================
# 8. Pie Chart
# ===============================

```python
labels = ["Python","Java","C++","JS"]
sizes = [40,25,20,15]

plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

# ===============================
# 9. Box Plot
# ===============================

```python
data = np.random.normal(50,10,100)

plt.figure(figsize=(6,4))
plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

# ===============================
# 10. Subplots
# ===============================

```python
x = np.arange(1,6)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(x, x**2)
plt.title("Square")

plt.subplot(1,2,2)
plt.plot(x, x**3)
plt.title("Cube")

plt.tight_layout()
plt.show()

print("✅ Day 21 Matplotlib Practice Completed")
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1,11)
y = x**2

# Figure Size
plt.figure(figsize=(8,5))

# Customized Line Plot
plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8,
    label="y = x²"
)

# Title & Labels
plt.title("Customized Line Plot", fontsize=16)
plt.xlabel("X Values", fontsize=12)
plt.ylabel("Y Values", fontsize=12)

# Grid
plt.grid(True)

# Legend
plt.legend()

# Axis Limits
plt.xlim(1,10)
plt.ylim(0,110)

# Annotation
plt.annotate(
    "Highest Point",
    xy=(10,100),
    xytext=(7,80),
    arrowprops=dict(arrowstyle="->")
)

# Save Figure
plt.savefig("custom_plot.png")

plt.show()
```
