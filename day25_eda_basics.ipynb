```python
import pandas as pd

# Load Dataset
df = pd.read_csv("Iris.csv")

# -----------------------
# Dataset Overview
# -----------------------

print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

# -----------------------
# Missing Values
# -----------------------

print("\nMissing Values:")
print(df.isnull().sum())

# -----------------------
# Duplicate Rows
# -----------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# -----------------------
# Summary Statistics
# -----------------------

print("\nSummary Statistics:")
print(df.describe())

# -----------------------
# Numerical Columns
# -----------------------

print("\nNumerical Columns:")
print(df.select_dtypes(include="number").columns.tolist())

# -----------------------
# Categorical Columns
# -----------------------

print("\nCategorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())

# -----------------------
# Unique Values
# -----------------------

for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("car_price.csv")
print(df.head())
```

# ---------------------------
# Price Distribution
# ---------------------------

```python
plt.figure(figsize=(8,5))
plt.figure(figsize=(8,5))
plt.hist(df["price"], bins=20, edgecolor="black")
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

# ---------------------------
# Fuel Type Distribution
# ---------------------------

```python
fuel = df["fueltype"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(fuel.index, fuel.values)
plt.title("Fuel Type Distribution")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.show()
```


# ---------------------------
# Transmission Distribution
# ---------------------------

```python
trans = df["drivewheel"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(trans.index, trans.values)
plt.title("Transmission Distribution")
plt.xlabel("Transmission")
plt.ylabel("Count")
plt.show()
```

# ---------------------------
# Seller Type Distribution
# ---------------------------

```python
seller_col = next(
    (c for c in ["Seller_Type", "seller_type", "SellerType", "sellerType"] if c in df.columns),
    None,
)

if seller_col is None:
    print("Column 'Seller_Type' not found in df. Available columns:", df.columns.tolist())
else:
    seller = df[seller_col].value_counts()
    plt.figure(figsize=(6,4))
    plt.bar(seller.index.astype(str), seller.values)
    plt.title("Seller Type Distribution")
    plt.xlabel("Seller Type")
    plt.ylabel("Count")
    plt.show()

plt.figure(figsize=(6,4))
if seller_col is not None:
    plt.bar(seller.index, seller.values)
else:
    plt.text(0.5, 0.5, "Column 'Seller_Type' not found", ha='center', va='center')
    plt.axis('off')
plt.title("Seller Type Distribution")
plt.xlabel("Seller Type")
plt.ylabel("Count")
plt.show()
```

# ---------------------------
# Owner Distribution
# ---------------------------


```python
owner_col = next(
    (c for c in ["Owner", "owner", "Ownership", "owner_type", "Owner_Type"] if c in df.columns),
    None,
)

if owner_col is None:
    print("Column 'Owner' not found in df. Available columns:", df.columns.tolist())
    owner = pd.Series(dtype=int)
else:
    owner = df[owner_col].value_counts()

plt.figure(figsize=(6,4))
plt.bar(owner.index.astype(str), owner.values)
plt.title("Ownership Distribution")
plt.xlabel("Owner")
plt.ylabel("Count")
plt.show()
```
