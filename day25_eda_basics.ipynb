```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("iris.csv")

print(df.head())
print(df.columns)
```

Sepal Length vs Sepal Width

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["SepalLengthCm"], df["SepalWidthCm"])

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.grid(True)

plt.show()
```

Petal Length vs Petal Width

```python
# ensure columns expected later in the cell exist (map from current CSV column names)
if "Species" in df.columns and "species" not in df.columns:
    df["species"] = df["Species"]
if "PetalLengthCm" in df.columns and "petal_length" not in df.columns:
    df["petal_length"] = df["PetalLengthCm"]
if "PetalWidthCm" in df.columns and "petal_width" not in df.columns:
    df["petal_width"] = df["PetalWidthCm"]

species = df["species"].unique()

for flower in species:

    data = df[df["species"] == flower]

    plt.scatter(
        data["petal_length"],
        data["petal_width"],
        label=flower
    )

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.grid(True)

plt.show()
```

Feature Relationship by Species

```python
species = df["species"].unique()

for flower in species:

    data = df[df["species"] == flower]

    plt.scatter(
        data["petal_length"],
        data["petal_width"],
        label=flower
    )

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.grid(True)

plt.show()
```

Average Features by Species

```python
if "sepal_length" not in df.columns and "SepalLengthCm" in df.columns:
    df["sepal_length"] = df["SepalLengthCm"]
if "sepal_width" not in df.columns and "SepalWidthCm" in df.columns:
    df["sepal_width"] = df["SepalWidthCm"]

avg_features = df.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].mean()

print(avg_features)

print(avg_features)
```

. Correlation Analysis

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

print(correlation)
```

Correlation Visualization

```python
plt.figure(figsize=(8, 6))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Iris Feature Correlation")

plt.tight_layout()
plt.show()
```

Strongest Correlation with Petal Width

```python
print(
    correlation["petal_width"]
    .sort_values(ascending=False)
)
```
