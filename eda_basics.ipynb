Petal Length vs Petal Width

```python
# ensure columns expected later in the cell exist (map from current CSV column names)
if "Species" in df.columns and "species" not in df.columns:
    df["species"] = df["Species"]
if "PetalLengthCm" in df.columns and "petal_length" not in df.columns:
    df["petal_length"] = df["PetalLengthCm"]
if "PetalWidthCm" in df.columns and "petal_width" not in df.columns:
    df["petal_width"] = df["PetalWidthCm"]

species = df["species"].unique()

for flower in species:

    data = df[df["species"] == flower]

    plt.scatter(
        data["petal_length"],
        data["petal_width"],
        label=flower
    )

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================
# Load Dataset
# ==========================

df = pd.read_csv("iris.csv")

print(df.head())
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())

# ==========================
# Select Numerical Features
# ==========================

features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

print("\nNumerical Features:")
features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

print("\nNumerical Features:")
print(df[features].describe())

species_mean = df.groupby("Species")[features].mean()
species = df["Species"].unique()

# later use:
for s in species:
    data = df[df["Species"] == s]

# ==========================
# Correlation Matrix
# ==========================

correlation = df[features].corr()

print("\nCorrelation Matrix:")
print(correlation)

# ==========================
# Correlation Heatmap
# ==========================

plt.figure(figsize=(8, 6))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(
    range(len(features)),
    features,
    rotation=45
)

plt.yticks(
    range(len(features)),
    features
)

plt.title("Iris Feature Correlation")

plt.tight_layout()
plt.show()

# ==========================
# Species-wise Mean
# ==========================

species_mean = df.groupby("Species")[features].mean()

print("\nAverage Features by Species:")
print(species_mean)

# ==========================
# Species-wise Feature Plot
# ==========================

species = df["Species"].unique()

for feature in features:

    plt.figure(figsize=(8, 5))

    values = [
        df[df["Species"] == s][feature].mean()
        for s in species
    ]

    plt.bar(species, values)

    plt.title(f"Average {feature} by Species")
    plt.xlabel("Species")
    plt.ylabel(f"Average {feature}")

    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

# ==========================
# Petal Features by Species
# ==========================

for s in species:

    for s in species:
        data = df[df["Species"] == s]
        plt.scatter(
            data["PetalLengthCm"],
            data["PetalWidthCm"],
            label=s
        )

    plt.scatter(
        data["PetalLengthCm"],
        data["PetalWidthCm"],
        label=s
    )

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.grid(True)

plt.show()

# ==========================
# Sepal Features by Species
# ==========================

for s in species:

    data = df[df["Species"] == s]

    plt.scatter(
        data["SepalLengthCm"],
        data["SepalWidthCm"],
        label=s
    )

plt.title("Sepal Length vs Sepal Width by Species")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print("Dataset:")
print(df.head())

# Numerical columns
features = iris.feature_names

# Check descriptive statistics
print("\nDescriptive Statistics:")
print(df[features].describe())

# Detect outliers using IQR
for column in features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Outliers:", len(outliers))

# Boxplots
plt.figure(figsize=(10, 6))

df[features].boxplot()

plt.title("Outlier Detection using Boxplots")
plt.ylabel("Value")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print("Original shape:", df.shape)

# -----------------------------
# 1. Check missing values
# -----------------------------

print("\nMissing values:")
print(df.isnull().sum())

# -----------------------------
# 2. Check duplicate rows
# -----------------------------

print("\nDuplicate rows:")
print(df.duplicated().sum())

# -----------------------------
# 3. Remove duplicates
# -----------------------------

df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)

# -----------------------------
# 4. Check data types
# -----------------------------

print("\nData types:")
print(df.dtypes)

# -----------------------------
# 5. Final missing-value check
# -----------------------------

print("\nFinal missing values:")
print(df.isnull().sum())

# -----------------------------
# 6. Final dataset info
# -----------------------------

print("\nCleaned dataset:")
print(df.info())

print("\nFirst 5 rows:")
print(df.head())
```
